# Comparing Gaussian bandit policies

```python
import numpy as np

from pymab.environments import BanditEnvironment
from pymab.policies import (
    EpsilonGreedyPolicy,
    GaussianThompsonSamplingPolicy,
    GreedyPolicy,
    SoftmaxPolicy,
    UCBPolicy,
)
from pymab.simulation import Experiment, ExperimentConfig

environment = BanditEnvironment(means=np.array([-0.2, 0.1, 0.6, 1.0]))
result = Experiment(
    environment=environment,
    policies={
        "greedy": GreedyPolicy(n_arms=4, initial_value=1.0),
        "epsilon-greedy": EpsilonGreedyPolicy(n_arms=4, epsilon=0.1),
        "softmax": SoftmaxPolicy(n_arms=4, temperature=0.5),
        "ucb": UCBPolicy(n_arms=4),
        "thompson": GaussianThompsonSamplingPolicy(n_arms=4),
    },
    config=ExperimentConfig(n_replicates=100, horizon=250, seed=5),
).run()

result.average_reward_by_step[-1]
```
