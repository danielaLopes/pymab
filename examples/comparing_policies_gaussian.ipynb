# Comparing Gaussian bandit policies

```python
import numpy as np

from pymab.environments import BanditEnvironment
from pymab.policies import (
    EpsilonGreedyPolicy,
    GaussianThompsonSamplingPolicy,
    GreedyPolicy,
    SoftmaxPolicy,
    UCBPolicy,
)
from pymab.simulation import Experiment, ExperimentConfig

environment = BanditEnvironment(q_values=np.array([-0.2, 0.1, 0.6, 1.0]))
result = Experiment(
    environment=environment,
    policies=[
        GreedyPolicy(n_arms=4, initial_value=1.0),
        EpsilonGreedyPolicy(n_arms=4, epsilon=0.1),
        SoftmaxPolicy(n_arms=4, temperature=0.5),
        UCBPolicy(n_arms=4),
        GaussianThompsonSamplingPolicy(n_arms=4),
    ],
    config=ExperimentConfig(n_episodes=100, n_steps=250, seed=5),
).run()

result.average_reward_by_step[-1]
```
